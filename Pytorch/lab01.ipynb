**Bài 8. Autograd cơ bản**

**Yêu cầu:**

• Tính thủ công đạo hàm của y = x2 tại x = 3.6 và so sánh với x.grad.

• Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.

```python
import torch

# 1. Tính thủ công đạo hàm của y = x^2 tại x = 3.6 và so sánh với x.grad
print("=== 1. Hàm y = x^2 tại x = 3.6 ===")
x1 = torch.tensor(3.6, requires_grad=True)
y1 = x1 * x1
y1.backward()

grad_autograd_1 = x1.grad.item()
grad_manual_1 = 2 * 3.6

print(f"y1 = {y1.item()}")
print(f"x1.grad (Autograd): {grad_autograd_1}")
print(f"Đạo hàm thủ công (2 * 3.6): {grad_manual_1}")
print(f"Khớp nhau: {abs(grad_autograd_1 - grad_manual_1) < 1e-5}")

# 2. Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient
print("=== 2. Đổi hàm thành y = 3*x^2 + 2*x + 1 tại x = 3.6 ===")
x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * (x2**2) + 2 * x2 + 1
y2.backward()

grad_autograd_2 = x2.grad.item()
grad_manual_2 = 6 * 3.6 + 2.0

print(f"y2 = {y2.item()}")
print(f"x2.grad (Autograd): {grad_autograd_2}")
print(f"Đạo hàm thủ công (6 * 3.6 + 2): {grad_manual_2}")
print(f"Khớp nhau: {abs(grad_autograd_2 - grad_manual_2) < 1e-5}")
```

**Bài 9. Gradient Descent tìm hệ số đa thức**

**Yêu cầu:**

• So sánh ba hệ số học được với [1, 2, 3].

• Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step().

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([X * X, X, np.ones_like(X)])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:", w.detach().numpy().flatten())

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

learned_coeffs = w.detach().numpy().flatten()
print("Learned coefficients:", learned_coeffs)

# 1. So sánh ba hệ số học được với [1, 2, 3]
print("=== 1. So sánh hệ số học được với [1, 2, 3] ===")
print("Hệ số thực tế target: [1.0, 2.0, 3.0]")
print("Hệ số học được:     ", learned_coeffs)
print("Độ chênh lệch absolute error:", np.abs(learned_coeffs - np.array([1, 2, 3])))

# 2. Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step()
print("=== 2. Giải thích vai trò của các hàm ===")
print("- optimizer.zero_grad(): Đặt gradient của các tham số (w) về 0 trước bước tính mới, tránh cộng dồn gradient từ epoch trước.")
print("- mse.backward(): Thực hiện lan truyền ngược (backpropagation) tính đạo hàm của loss theo w và lưu vào w.grad.")
print("- optimizer.step(): Cập nhật trọng số w theo gradient vừa tính và thuật toán NAdam.")
```

**Bài 10. Giải hệ phương trình bằng Autograd**

**Yêu cầu:**

• Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.

• Nhận xét vì sao bài toán có thể có nhiều nghiệm.

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("Nghiệm tìm được:")
print("A =", A.item())
print("B =", B.item())
print("C =", C.item())
print("D =", D.item())

# 1. Kiểm tra nghiệm thu được với bốn phương trình
print("=== 1. Kiểm tra nghiệm thu được ===")
eq1 = A.item() + B.item()
eq2 = C.item() - D.item()
eq3 = A.item() + C.item()
eq4 = B.item() - D.item()

print(f"PT 1: A + B = {eq1:.4f} (Target: 9)")
print(f"PT 2: C - D = {eq2:.4f} (Target: 1)")
print(f"PT 3: A + C = {eq3:.4f} (Target: 8)")
print(f"PT 4: B - D = {eq4:.4f} (Target: 2)")
print(f"Tổng sai số bình phương (Loss): {sqerr.item():.6f}")

# 2. Nhận xét vì sao bài toán có thể có nhiều nghiệm
print("=== 2. Nhận xét vì sao bài toán có nhiều nghiệm (vô số nghiệm) ===")
print("- Hệ phương trình có 4 ẩn (A, B, C, D) nhưng các phương trình bị phụ thuộc tuyến tính:")
print("  (A + B = 9) và (A + C = 8) => B - C = 1.")
print("  Kết hợp (C - D = 1) => B - D = 2 (trùng với PT 4).")
print("- Ma trận hệ số có rank = 3 < 4 (số ẩn), do đó hệ có vô số nghiệm. Thuật toán Gradient Descent sẽ hội tụ về nghiệm tùy thuộc vào điểm khởi tạo ban đầu của các biến.")
```

**Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau:**

• Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4 × 6, 2 × 12 và 2 × 3 × 4.

• Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.

• Cho y = 2x2 + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.

• Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.


```python
import torch
import numpy as np

# -------------------------------------------------------------------------
# 1. Tạo Tensor X có 24 phần tử và reshape lần lượt
# -------------------------------------------------------------------------
print("=== Yêu cầu 1: Reshape Tensor X ===")
X = torch.arange(1, 25, dtype=torch.float32)
print("Tensor X ban đầu (24 phần tử):", X)

X_4x6 = X.reshape(4, 6)
print("\nReshape thành 4x6:\n", X_4x6)

X_2x12 = X.reshape(2, 12)
print("\nReshape thành 2x12:\n", X_2x12)

X_2x3x4 = X.reshape(2, 3, 4)
print("\nReshape thành 2x3x4:\n", X_2x3x4)


# -------------------------------------------------------------------------
# 2. Phép toán trên hai Tensor A, B (kích thước 3x4)
# -------------------------------------------------------------------------
print("\n=== Yêu cầu 2: Phép toán trên Tensor A, B ===")
A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("Tensor A:\n", A)
print("Tensor B:\n", B)

print("\nA + B:\n", A + B)
print("A * B (Hadamard product):\n", A * B)
print("A / B (Element-wise division):\n", A / B)

# Trung bình và độ lệch chuẩn theo cột (dim=0)
print("\nTrung bình theo cột của A:\n", torch.mean(A, dim=0))
print("Độ lệch chuẩn theo cột của A:\n", torch.std(A, dim=0))
print("Trung bình theo cột của B:\n", torch.mean(B, dim=0))
print("Độ lệch chuẩn theo cột của B:\n", torch.std(B, dim=0))


# -------------------------------------------------------------------------
# 3. Tính dy/dx của y = 2x^2 + 5x + 3 tại x = 2 với autograd & kiểm tra thủ công
# -------------------------------------------------------------------------
print("\n=== Yêu cầu 3: Autograd tính đạo hàm ===")
x = torch.tensor(2.0, requires_grad=True)
y = 2 * (x**2) + 5 * x + 3
y.backward()

grad_autograd = x.grad.item()
# Đạo hàm thủ công: dy/dx = 4x + 5. Tại x = 2: 4*(2) + 5 = 13.0
grad_manual = 4 * 2.0 + 5.0

print(f"Giá trị y tại x = 2: {y.item()}")
print(f"Đạo hàm dy/dx tính bởi Autograd: {grad_autograd}")
print(f"Đạo hàm dy/dx tính thủ công: {grad_manual}")
print(f"Kết quả trùng khớp: {np.isclose(grad_autograd, grad_manual)}")


# -------------------------------------------------------------------------
# 4. Huấn luyện Gradient Descent tìm hệ số đa thức y = 2x^2 - 3x + 5
# -------------------------------------------------------------------------
print("\n=== Yêu cầu 4: Gradient Descent tìm hệ số y = 2x^2 - 3x + 5 ===")
# Tạo dữ liệu giả định theo đa thức y = 2x^2 - 3x + 5
polynomial_target = np.poly1d([2, -3, 5])
N = 100
X_np = np.random.randn(N, 1) * 5
Y_np = polynomial_target(X_np)

# Xây dựng ma trận đặc trưng [x^2, x, 1]
XX_np = np.hstack([X_np**2, X_np, np.ones_like(X_np)])

x_tensor = torch.tensor(XX_np, dtype=torch.float32)
y_tensor = torch.tensor(Y_np, dtype=torch.float32)

# Khởi tạo tham số w ngẫu nhiên
w = torch.randn(3, 1, requires_grad=True)
optimizer = torch.optim.NAdam([w], lr=0.01)

print("Hệ số khởi tạo ban đầu:\n", w.detach().numpy().flatten())

# Vòng lặp huấn luyện
EPOCHS = 2000
for epoch in range(EPOCHS):
    optimizer.zero_grad()
    y_pred = x_tensor @ w
    mse = torch.mean((y_tensor - y_pred) ** 2)
    mse.backward()
    optimizer.step()

learned_coeffs = w.detach().numpy().flatten()
print(f"\nHệ số học được sau {EPOCHS} epochs: {learned_coeffs}")
print("Hệ số thực tế target [2, -3, 5]: [2.0, -3.0, 5.0]")
```
